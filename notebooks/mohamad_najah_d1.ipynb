# Data Cleaning

```python
#count missing values per column 
df.isnull().sum()
```

```python
df.isnull().sum().sum()
```

```python
#express missingness as a propotion 
df.isnull().mean()
```

```python
# inspect the affected rows 
df[df['physical_activity'].isnull()]
```

```python
df[df.isnull().any(axis=1)]
```

```python
#count the placeholders 
(df['skin_thickness'] == 0).sum()
```

```python
(df['glucose'] == 0).sum()
```

```python
placeholder_cols = ['glucose','blood_pressure','skin_thickness','insulin','bmi']
for i in placeholder_cols:
    print(i,'->',(df[i] == 0).sum(),'zeros')
```

```python
import numpy as np
```

```python
#convert placeholders to missing values
df['glucose'] = df['glucose'].replace(0, np.nan)
```

```python
before_missing = df.isnull().sum().sum()

for i in placeholder_cols:
    df[i] = df[i].replace(0,np.nan)

after_missing = df.isnull().sum().sum()
print('missing cells before:', before_missing)
print('missing cells before:', after_missing)
```

```python
#re-measure the damage 
df.isnull().sum()
```

```python
df.isnull().mean()
```

```python
# fill numerical columns
df['glucose'] = df['glucose'].fillna(df['glucose'].mean())
```

```python
column_fill = ['age','blood_pressure','skin_thickness','insulin','bmi']
for i in column_fill:
    df[i] = df[i].fillna(df[i].median())

```

```python
df.isnull().sum()
```

```python
#fill categorical values
catogerical_fill = ['physical_activity','smoker']
for i in catogerical_fill:
    df[i] = df[i].fillna('Unknown')
df[i].isnull().sum()
```

```python
#verify no missing values remain
df.isnull().sum().sum()
```

```python
df['gender'].unique()
```

```python
#remove stray whitespace
df['gender'] = df['gender'].str.strip()
```

```python
df['gender'].unique()
```

```python
df.value_counts('gender')
```

```python
for i in catogerical_fill:
    df[i] = df[i].str.strip()
```

```python
df.value_counts(['physical_activity','smoker'])
```

```python
df['physical_activity'].unique()
```

```python
#standardize capitalization 
df['gender'] = df['gender'].str.lower()
```

```python
df.value_counts('gender')
```

```python
df['physical_activity'] = df['physical_activity'].str.title()
```

```python
df['physical_activity'].value_counts()
```

```python
df['smoker'] = df['smoker'].str.upper()
```

```python
df.value_counts('smoker')
```

```python
#unify equivalent categories 
df['gender'] = df['gender'].replace({'f': 'female','m': 'male'})
```

```python
df.value_counts('gender')
```

```python
df['smoker'] = df['smoker'].replace({'N': 'NO','Y': 'YES'})
```

```python
df.value_counts('smoker')
```

```python
#convert text dates to real dates
df['last_checkup_date'] = pd.to_datetime(df['last_checkup_date'],dayfirst=True)
```

```python
df.dtypes
```

```python
df.value_counts('last_checkup_date')
```

```python
#correct remaining type mismatches 
df['age'] = df['age'].round().astype(int)
```

```python
df.value_counts('age')
```
